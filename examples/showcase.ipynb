ثلاث هويات في دفتر واحد. المظهر معزول لكل deck. افتح showcase.html لمعرض مستقل جاهز.

```python
from lecture_studio import LectureDeck, Theme

gold = LectureDeck(course="Lecture Studio · معرض المكونات", series="DESIGN SYSTEM")
gold.cover("مادة واحدة. هوية متكاملة.", "التصميم نفسه يخدم جميع موادك.\nغيّر المحتوى والهوية، واحتفظ بتجربة تعلم واضحة.", tags=["ديناميكي", "قابل لإعادة الاستخدام"], hero_code="content → design → learning", hero_result="Ready to teach")
```

```python
for theme in ["gold", "emerald", "ocean"]:
    demo = LectureDeck(course="ثيم " + theme, theme=theme)
    demo.slide("محتوى مختلف، تجربة متسقة", cards=[{"title": "الفكرة", "explanation": "تكتب المحتوى فقط."}, {"title": "تطبيق", "code": "print(42)"}])
```

```python
gold.slide("من المفهوم إلى التطبيق", diagram={"title": "مسار التعلم", "direction": "rtl", "nodes": [{"value": "افهم"}, {"value": "توقّع"}, {"value": "جرّب", "selected": True}, {"value": "فسّر"}]}, activity={"question": "كيف تتأكد من الفهم؟", "answer": "غيّر المثال وفسّر النتيجة."})
```

```python
gold.table("المكونات", ["المكون", "الوظيفة"], [["cover", "غلاف"], ["slide", "شرح"], ["diagram", "رسم"], ["activity", "سؤال"]])
gold.note("التخصيص", "تستطيع تغيير الألوان والاتجاه والحركة.")
```
